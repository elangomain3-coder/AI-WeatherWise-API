<h3>Handling null values of Age column</h3>

```python
df['Age'].describe()
```

```python
df['Age']=df['Age'].fillna(df['Age'].median())
```

The Age column contained approximately 15% missing values. To handle these missing values, median imputation was used. The median value was chosen instead of the mean because it is less affected by extreme or unusual values that may exist in healthcare data. Descriptive statistics showed that the age distribution was fairly balanced, but using the median provides a more robust and reliable representation of the central age group. This helps maintain the natural distribution of patient ages and reduces the impact of potential outliers.

<h3>Handling null values of gender column</h3>

```python
df['Gender'].value_counts()
```

```python
df['Gender']=df['Gender'].fillna('not_specified')
```

The Gender column contained missing values. Since gender distribution was nearly balanced, mode imputation could introduce slight bias. Therefore, missing values were replaced with "Unknown" to maintain data transparency and avoid assumptions about patient gender.

<h3>Handling null values of Medical condition column</h3>

```python
df['Medical Condition'].value_counts()
```

```python
df['Medical Condition']=df['Medical Condition'].fillna('Unknown')
```

```python
df[df['Medical Condition']=='Unknown']
```

Missing values in the Medical Condition column were replaced with "Not Specified" instead of using the mode. Since medical conditions are sensitive health attributes, replacing missing values with the most frequent condition could introduce bias and misrepresent patient health status. Using "Not Specified" preserves data integrity and allows better interpretation during exploratory analysis.

<h3>Handling null values of Glucose & Blood Pressure  column</h3>

```python
null_cols=['Glucose','Blood Pressure']
```

```python
for col in null_cols:
    df[col]=df[col].fillna(df[col].median())
```

Missing values in Glucose and Blood Pressure columns were filled using median values. Median was chosen because these medical measurements may contain extreme values, and median is less affected by outliers. This helped maintain realistic patient health data and ensured the dataset was ready for further analysis.

<h2>Detecting Outliers </h2>

```python
def detect_outlier(DataFrame:df,col):
    for i in col:
        Q1=df[i].quantile(0.25)
        Q3=df[i].quantile(0.75)
        IQR=Q3-Q1
        lw=Q1-1.5*IQR
        uw=Q3+1.5*IQR
        outliers=df[(df[i]<lw)|(df[i]>uw)]
     
        print(f"{i}")
        print(f"{i}->lw-{lw}")
        print(f"{i}->uw-{uw}")
        print(f"total outlier in {i}-> {len(outliers)}")
        print("--------------------------"*5)
```

<pre><h5>glucose</h5>
Risk Level     Glucose 
Normal          70–99
Prediabetic     100–125
Diabetic         ≥126

```python
def glucose_risk(x):
    if x > 500:
        return 'Extreme Outlier'
    elif x >= 126:
        return 'Diabetic'
    elif x >= 100:
        return 'Prediabetic'
    else:
        return 'Normal'

df['Glucose_Risk'] = df['Glucose'].apply(glucose_risk)
```

The Lifestyle_Index is a single score that shows how healthy a person’s lifestyle is. It adds together their physical activity, diet, and sleep, and subtracts stress. A higher score means a healthier lifestyle, while a lower score shows more unhealthy habits. This makes it easy to compare people and see who may need to improve their lifestyle.
