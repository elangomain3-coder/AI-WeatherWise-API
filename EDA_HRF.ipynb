<h2>Problem Statement & Business Objectives</h2>
Healthcare organizations need to understand how different health factors such as age, lifestyle habits, and medical measurements influence diseases like diabetes, hypertension, and obesity. This dataset contains patient health information that helps analyze these risk factors and identify patterns related to various medical conditions.
The main objective of this analysis is to explore the dataset to identify important health factors linked to diseases. The study focuses on understanding how lifestyle habits such as smoking, alcohol consumption, diet, sleep, and physical activity affect health. It also aims to analyze how demographic factors like age and gender influence disease occurrence. Additionally, the analysis helps identify relationships between medical indicators such as glucose level, BMI, blood pressure, and cholesterol. The insights gained from this study can support early disease detection and promote preventive healthcare strategies.

### Dataset Summary:
- Contains healthcare-related patient data combining clinical, lifestyle, and risk information
- Includes physiological measurements like BMI, glucose, blood pressure, cholesterol, triglycerides, HbA1c, and oxygen saturation
- Covers lifestyle behavior factors such as physical activity, diet score, stress level, sleep hours, and a Lifestyle Index
- Provides demographic information including age and gender
- Includes health risk categories like BMI category, glucose risk, and blood pressure category
- Contains medical outcome indicators such as medical condition and length of hospital stay
- Tracks family history of disease as a risk factor

```python
# Smoking and Alcohol affect BMI, Blood Pressure, and Glucose

# Calculate averages
smoking_avg = df.groupby('Smoking')[['BMI', 'Blood Pressure', 'Glucose']].mean()
alcohol_avg = df.groupby('Alcohol')[['BMI', 'Blood Pressure', 'Glucose']].mean()

# Create subplot layout (2 rows, 3 columns)
fig, axes = plt.subplots(2, 3, figsize=(18, 8))


# ----- Smoking Effects -----

smoking_avg['BMI'].plot(kind='bar', ax=axes[0,0],color='skyblue')
axes[0,0].set_title('Smoking vs BMI')
axes[0,0].set_xlabel('Smoking Status')
axes[0,0].set_ylabel('Average BMI')
axes[0,0].tick_params(axis='x',rotation=0)

smoking_avg['Blood Pressure'].plot(kind='bar', ax=axes[0,1],color='skyblue')
axes[0,1].set_title('Smoking vs Blood Pressure')
axes[0,1].set_xlabel('Smoking Status')
axes[0,1].set_ylabel('Average Blood Pressure')
axes[0,1].tick_params(axis='x',rotation=0)

smoking_avg['Glucose'].plot(kind='bar', ax=axes[0,2],color='skyblue')
axes[0,2].set_title('Smoking vs Glucose')
axes[0,2].set_xlabel('Smoking Status')
axes[0,2].set_ylabel('Average Glucose')
axes[0,2].tick_params(axis='x',rotation=0)


# ----- Alcohol Effects -----

alcohol_avg['BMI'].plot(kind='bar', ax=axes[1,0],color='skyblue')
axes[1,0].set_title('Alcohol vs BMI')
axes[1,0].set_xlabel('Alcohol Consumption')
axes[1,0].set_ylabel('Average BMI')
axes[1,0].tick_params(axis='x',rotation=0)

alcohol_avg['Blood Pressure'].plot(kind='bar', ax=axes[1,1],color='skyblue')
axes[1,1].set_title('Alcohol vs Blood Pressure')
axes[1,1].set_xlabel('Alcohol Consumption')
axes[1,1].set_ylabel('Average Blood Pressure')
axes[1,1].tick_params(axis='x',rotation=0)

alcohol_avg['Glucose'].plot(kind='bar', ax=axes[1,2],color='skyblue')
axes[1,2].set_title('Alcohol vs Glucose')
axes[1,2].set_xlabel('Alcohol Consumption')
axes[1,2].set_ylabel('Average Glucose')
axes[1,2].tick_params(axis='x',rotation=0)

plt.tight_layout()
plt.show()
```

The visuals indicate that both smokers and alcohol consumers have higher average BMI, blood pressure, and glucose levels compared to non-users. For smoking, the bar charts show a clear increase in average blood pressure and glucose, while the difference in BMI is present but moderate. In the case of alcohol consumption, individuals who consume alcohol display consistently higher averages across BMI, blood pressure, and glucose than non-consumers. Overall, the patterns across all six visuals suggest that smoking and alcohol consumption are associated with poorer health indicators, particularly in terms of blood pressure and glucose levels.

The correlation analysis reveals a clear metabolic risk cluster where BMI, glucose, HbA1c, cholesterol, and triglycerides are positively associated, indicating interrelated cardiometabolic risk factors. Lifestyle-related variables such as physical activity, diet score, sleep hours, and the composite lifestyle index show inverse relationships with these clinical markers, supporting their protective role. Length of stay demonstrates positive correlations with metabolic and stress indicators, suggesting that poorer health profiles are associated with prolonged hospitalization. Strong correlations between glucose and HbA1c, as well as among lipid measures, indicate potential multicollinearity that should be considered during predictive modeling.
